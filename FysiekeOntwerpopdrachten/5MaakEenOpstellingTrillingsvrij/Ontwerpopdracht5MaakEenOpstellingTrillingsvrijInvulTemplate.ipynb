```python
f = 5 # Hz
w = 2*np.pi*f # Hz
w0 = w/2
f0 = w0/(2*np.pi)
print(f"de kantelfrequentie is {f0:.1f} Hz")
```

```python
f = np.linspace(0,20,100)
A = np.abs(1/(1-((2*np.pi*f)**2/(w0**2))))

plt.figure
plt.loglog(f,A)
plt.xlabel("$log(f_v)(Hz)$")
plt.ylabel("log($A_m/A_v)$")
plt.show()
```

```python
g = 9.81 #m/s^2
d_x = g/(w0)**2
print(f"minimale waarde voor delta x= {d_x:.3f} m")
```

## Opdracht 3: Plot de grafiek.

```python
m = np.linspace(0,5,100)
def C(m,g,d_x):
    return (m*g)/d_x

plt.plot(m,C(m,g,d_x),label = "$\Delta x$")
plt.plot(m,C(m,g,d_x*2),label = "$\Delta x$*2")
plt.plot(m,C(m,g,d_x*5),label= "$\Delta x$*5")
plt.xlabel("m(kg)")
plt.ylabel("C(N/m)")
plt.legend()
plt.savefig("d_x_grafiek")
plt.show

print(C(0.5,g,d_x*1.5))
```
